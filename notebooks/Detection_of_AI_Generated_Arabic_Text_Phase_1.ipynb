```python
# Import data preparation functions
import data_preparation as dp
```

```python
# Load the dataset
dataset_name = "KFUPM-JRCAI/arabic-generated-abstracts"

dataset = dp.load_data(dataset_name)

print(dataset)
```

```python
# Inspect the structure of each subset
for split in dataset:

    print(f"\n\033[1mSubset: {split}\033[0m")

    dp.inspect_structure(dataset[split])
```

```python
# Convert each subset to a pandas DataFrame
subset = {}

for split in dataset:
    subset[split] = dp.to_dataframe(dataset[split])

print("Subset:", list(subset.keys()))
```

```python
# Define the original and comparison columns
original_column = "original_abstract"

generated_columns = [
    "allam_generated_abstract",
    "jais_generated_abstract",
    "llama_generated_abstract",
    "openai_generated_abstract"
]
```

```python
# Check the data types of each subset
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    dp.check_data_types(data)
```

```python
# Display the first 5 rows of each subset
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    display(dp.preview_data(data))
```

```python
# Check whether the label column exists
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    result = dp.check_column(data, "label")

    if result is None:
        print("No label column found")
    else:
        print(result)
```

```python
# Check the distribution of human and AI-generated texts
for split, data in subset.items():

    human_texts, ai_texts = dp.count_class_distribution(
        data,
        original_column,
        generated_columns
    )

    print(f"\n\033[1mSubset: {split}\033[0m")
    print("Human:", human_texts)
    print("AI-generated:", ai_texts)
```

```python
# Check for empty or whitespace-only cells
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    empty_values = dp.check_empty_values(data)

    for column, count in empty_values.items():
        print(f"{column}: {count} empty values")
```

```python
# Check for missing values
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    print(dp.check_missing_values(data))
```

```python
# Check for duplicate rows
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    duplicates = dp.check_duplicates(data)

    print("Duplicates:", duplicates)
```

```python
# Check the number of unique values in each column
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    unique_values = dp.check_unique_values(data)

    for column, count in unique_values.items():
        print(f"{column}: {count} unique values")
```

```python
# Check exact matches between the original and generated columns
for split, data in subset.items():

    print(f"\n\033[1mSubset: {split}\033[0m")

    exact_matches = dp.check_exact_matches(
        data,
        original_column,
        generated_columns
    )

    for column, count in exact_matches.items():
        print(f"{column}: {count} exact matches")
```

```python
# Check for additional numbers in the by_polishing subset
data = subset["by_polishing"]

print(f"\n\033[1mSubset: by_polishing\033[0m")

additional_numbers = dp.check_additional_numbers(
    data,
    original_column,
    generated_columns
)

for column, count in additional_numbers.items():
    print(f"{column}: {count} texts with additional numbers")
```
